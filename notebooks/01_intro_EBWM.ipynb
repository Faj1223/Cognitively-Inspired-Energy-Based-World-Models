## Qu'est ce qu'un EBWM ?
EBWM veut dire Energie Based World Model. Un EBWM est donc un Modele du Monde Basé sur l'énergie. Le concepte mélange deux notions majeures : les Worlds models et les Energie Base Models ( EBT ).

# World model et contexte

Avant l'émergence des World Models, le Renforcement Learning reposait éssentiellement sur des méthodes bien connues comme le Q-learning, Deep Q-Networks ou PPO. Dans ces méthodes classiques, l'agent qui est sensé apprendre, perçoit un état, choisit une action de manière quasi-aveugle et reçoit une récompense du milieu extérieur. L'apprentissage se fait à force d'itérations brutes. L'agent doit tout bonnement se fier à ce système de récompense pour comprendre par exemple que tomber d'une falaise est fatal. Dans les faits l'agent doit virtuellement tomber des milliers de fois pour comprendre celà. L'humain, vous et moi ne sommes jamais tomber d'une falaise sinon nous ne seront plus là. Et pourtant nous avons bien cerné qu'une chute depuis une falaise serait ftale.
Regardons le problème de plus prêt.

Un nouveau né, à peine les yeux ouverts à déjà passé des secondes à observer et à se faire une idée personnelle et interne du monde qui l'entoure. Il regarde et ajuste. L'une des premières chose qu'il sait faire c'est reconnaître le visage et la voix de sa mère en l'observant et en l'écoutant. Et tout ceci sans instruction extérieure. Il observe et ajuste. De même un jeune humain peut apprendre à conduire une voiture en seulement quelques heures de pratiques sans provoquer des milliers d'accidents. Un modèle artificielle basé sur le RL a besoin de plusieurs millions d'heures d'apprentissage et de millions d'erreurs. Un modèle de language ( LLM )peut engloutir des trillions de tokens de texte et pourtant être incapable de comprendre un sens commun physique. un trillions de tokens de texte, c'est plus de données textuelles qu'un humain ne lira jamais même s'il disposait de plusieurs vies. C'est en partant de ce constat que YANN LeCun introduit les Worl Models. Le cerveau humain n'apprends pas en obtenant des recompences. Il affine une représentation interne du monde qu'il se construit au fil du temps.

## Modélisation
On distingue deux types de World Models selon la manière dont ils prédisent le futur :

Les **modèles autorégressifs** ( step by step ) et les **Non-Autorégressifs / Joint-Embedding**, ( ex : JEPA ).
Dans ce projet il ne sera question que des modèles autorgressifs.

## World model autorégressif
Ils prédisent l'état futur s(t+1) de façon séquentielle, en réinjectant chaque prédiction comme entrée pour le pas suivant.

Fonctionnement :
$$s(t + 1) = F(s(t), x(t), a(t), z(t))$$

où $x(t)$ est une observation, $s(t)$ est une estimation précédente de l'état du monde, $a(t)$ est une action, $z(t)$ est une variable latente, et $F$ est la fonction utilisée.

## Energie Based Model 

Les modèles génératifs classiques cherchent à apprendre une distribution de probabilité conditionnelle $P(y|x)$ sous la forme d'une prédiction $y = f(x)$. Chose importante, le $y$ est unique, et pourtant plusieurs futurs $y_1, y_2, y_3$ peuvent être valides. Si l'ont force un réseau de neuronne ( par exemple ) à prédire une seule valeur avec une perte de type MSE, alors le modèle pourrait tout à fait prédire un la moyenne des futurs possibles, et celà peut parfois prendre une forme physique absurde. 
Par exemple dans le cas de la conduite autonome, une voiture qui doit passer entre deux camions à tout un ensemble de solution sous forme de bande. L'ensemble solution est continue. Un modèle de deep learning prédirait la moyenne qui est donc le parfait milieu de la bande de traverséé. Celà peut être très couteux...
Un autre exemple, nous disposons d'une image bruitée. Le bruit est trafiqué de sorte à ce que l'image ressemble à un chat mais aussi à un chien. Plusieurs sorties sont donc plausibles. Un chat ou un chien.
ça, un **EBM** pourrait le faire, mais un modele de prédiction unique peut chercher à moyenner ces deux sorties ce qui donnerait un une image qui ne ressemble ni à un chien ni à un chat.
On introduit donc les **EBM** dont le principe consiste à produit sachant une donnée une entrée **x**, une ou plusieurs sortie(s) **y** pour la(es)quelle(s) l'énergie **E(x,y)** est minimale ou en dessous d'un certain seuil.


Dans ce article les auteurs mettent l'accent sur deux approches autoregressives traditionnelles.
Les **Transformers Auto Regressifs** et les **Réseau de neurones récurrents**.
Un **Transformer** est une architcteure de réseau de neurones introduite en 2017 par une équipe de huit chercheurs de Google Brain et Google Research dans l'article de référence **Attention Is All You Need**.
La particularité d'un **Transformer** est que son l'architecture repose éssentiellement sur l'association du **mécanisme d'attention** et de la mise en place de calculs matriciels parallèles. 
Quand au **Transformer Auto Regressifs** ou **ART** c'est donc un **Transformer** qui génère des séquences (comme du texte (token) par exemple dans le cas des LLM) pas à pas en prédisant l'élément suivant à partir de l'historique des éléments précédemment prédits.
